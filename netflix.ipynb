```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

**Import dataset**

```python
import pandas as pd

#  Added encoding='latin-1' to bypass the UTF-8 decode error
df = pd.read_csv('/content/netflix_dataset.csv', header=None, encoding='latin-1')

transaction = []
for i in range(0, 7466):
    transaction.append([str(df.values[i, j]) for j in range(0, 20)])
```

**Eclat Training on dataset**

```python
import apyori
rules = apyori.apriori(transaction, min_support=0.003, min_confidence=0.2, min_lift=3, min_length=2, max_length=2)
```

**Visulizing**

```python
result = list(rules)
result
```

**Proper Display**

```python
def inspect(result):
    movie_1 = [tuple(result[2][0][0])[0] for result in result]
    movie_2 = [tuple(result[2][0][1])[0] for result in result]
    support = [result[1] for result in result]
    return list(zip(movie_1, movie_2, support))
result_in_dataframe = pd.DataFrame(inspect(result), columns=['Movie 1', 'Movie 2', 'Support'])
```
